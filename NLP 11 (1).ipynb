```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# ------------------------------------------------------------
# 1. Prepare the sample sentiment dataset
# ------------------------------------------------------------

samples = [
    ("I love this movie", "positive"),
    ("This movie is amazing", "positive"),
    ("The movie was excellent", "positive"),
    ("I really enjoyed this film", "positive"),
    ("This is a wonderful movie", "positive"),
    ("I hate this movie", "negative"),
    ("This movie is terrible", "negative"),
    ("The movie was boring", "negative"),
    ("I did not like this film", "negative"),
    ("This is a bad movie", "negative")
]

dataset = pd.DataFrame(
    samples,
    columns=["review", "sentiment"]
)

print("Sample Dataset")
print(dataset.to_string(index=False))


# ------------------------------------------------------------
# 2. Separate input text and target labels
# ------------------------------------------------------------

texts = dataset["review"]
labels = dataset["sentiment"]

train_text, test_text, train_label, test_label = train_test_split(
    texts,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)


# ------------------------------------------------------------
# 3. Create the NLP classification pipeline
# ------------------------------------------------------------

sentiment_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            lowercase=True,
            stop_words=None
        )
    ),
    (
        "classifier",
        SVC(kernel="linear")
    )
])


# ------------------------------------------------------------
# 4. Train the model
# ------------------------------------------------------------

sentiment_model.fit(
    train_text,
    train_label
)


# ------------------------------------------------------------
# 5. Evaluate the model
# ------------------------------------------------------------

test_predictions = sentiment_model.predict(test_text)

print("\nActual Labels:")
print(test_label.to_numpy())

print("\nPredicted Labels:")
print(test_predictions)

score = accuracy_score(
    test_label,
    test_predictions
)

print(f"\nModel Accuracy: {score:.2f}")

print("\nClassification Report:")
print(
    classification_report(
        test_label,
        test_predictions,
        zero_division=0
    )
)


# ------------------------------------------------------------
# 6. Test the model with new sentences
# ------------------------------------------------------------

examples = pd.Series([
    "I really love this movie",
    "This movie is very boring"
])

example_predictions = sentiment_model.predict(examples)


# Display predictions in a table
results = pd.DataFrame({
    "Input": examples,
    "Prediction": example_predictions
})

print("\nPredictions for New Sentences:")
print(results.to_string(index=False))
```
